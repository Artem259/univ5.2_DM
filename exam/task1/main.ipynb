```python
import pandas as pd
from mlxtend.classifier import OneRClassifier as mlxtend_OneRClassifier

import service
from classification.OneRClassifier import OneRClassifier
```

```python
df = pd.read_csv('data/data_3x.csv')  # TODO change to variant
df.info()
```

```python
X, y = df.drop(columns='y'), df['y']
X, y = X.values, y.values
X.shape, y.shape
```

```python
X_pred = pd.DataFrame({'x0': [1], 'x1': [1], 'x2': [1]})  # TODO change to variant
X_pred = X_pred.values
```

#### OneRClassifier

```python
clf = OneRClassifier()
clf.fit(X, y);
```

```python
service.OneRClassifier_info(clf)
```

```python
y_pred = clf.predict(X_pred)
y_pred
```

Compare with similar `mlxtend` model

```python
mlxtend_clf = mlxtend_OneRClassifier()
mlxtend_clf.fit(X, y);
```

```python
mlxtend_y_pred = mlxtend_clf.predict(X_pred)
mlxtend_y_pred
```

```python
assert y_pred == mlxtend_y_pred
```
