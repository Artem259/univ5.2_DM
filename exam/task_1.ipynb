```python
import numpy as np
import pandas as pd
```

```python
data = [5, 7, 6, 8, 7, 9]
window_size = 2
step = 1
```

```python
def custom_sliding_window_avg(data, window_size, step):
    result = []
    for i in range(0, len(data) - window_size + 1, step):
        window = data[i:i + window_size]
        result.append(sum(window) / window_size)
    return result

def pandas_sliding_window_avg(data, window_size, step):
    series = pd.Series(data)
    rolling_means = series.rolling(window=window_size).mean()
    return rolling_means[window_size - 1::step].tolist()
```

```python
custom_result = custom_sliding_window_avg(data, window_size, step)
pandas_result = pandas_sliding_window_avg(data, window_size, step)
```

```python
print("Custom result: ", custom_result)
```

```python
print("Pandas result:", pandas_result)
```

```python
print("Equal?", np.allclose(custom_result, pandas_result, equal_nan=True))
```
